# AI Web Summarizer using Local Models

## Importing the libraries

```python
import os
from openai import OpenAI
from google import genai
from dotenv import load_dotenv
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from IPython.display import Markdown, display
```

## Loading Gemini API Key

```python
load_dotenv(override = True)
api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    print("No API key found!")
else:
    print("API key found!")
```

## Calling through http request

```python
load_dotenv(override = True)
openai_api_key = os.getenv("OPENAI_API_KEY")

if not openai_api_key:
    print("No API key found!")
else:
    print("API key found!")
```

```python
import requests
message = "Hey ChatGPT!, Find x when x + 10 = 20x?"

headers = {
    "Authorization": f"Bearer {openai_api_key}",
    "Content-Type": "application/json"
}

payload = {
    "model": "gpt-5.4-mini",
    "messages": [
        {
            "role": "system",
            "content": "Act like a math teacher"
        },

        {
            "role": "user",
            "content": message
        }  
    ]
}
```

```python
OPENAI_BASE_URL = "https://api.openai.com/v1/chat/completions"

response = requests.post(
    OPENAI_BASE_URL,
    headers = headers,
    json = payload
)
```

```python
response.json()
```

```python
response.json()["choices"][0]["message"]["content"]
```

## Calling through API

```python
message = "Hey Gemini!, Find x when x + 10 = 20x?"

client = genai.Client()

interaction = client.interactions.create(
    model = "gemini-3.1-flash-lite",
    system_instruction = "Act like a math teacher",
    input = message
)

print(interaction.output_text)
```

## Calling Gemini through OpenAI

```python
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
message = "Hey Gemini!, Find x when x + 10 = 20x?"

client = OpenAI(
    base_url = GEMINI_BASE_URL,
    api_key = api_key
)

response = client.chat.completions.create(
    model = "models/gemini-3.1-flash-lite",
    messages = [
        {
            "role": "system",
            "content": "Act like a math teacher"
        },

        {
            "role": "user",
            "content": message
        }  
    ]
)

print(response.choices[0].message.content)
```

## Calling local model through OpenAI

```python
LOCAL_BASE_URL = "http://localhost:11434/v1"

client = OpenAI(
    base_url = LOCAL_BASE_URL,
    api_key = "ollama"
)

response = client.responses.create(
    model = "qwen3:0.6b",
    input = [
        {
            "role": "system",
            "content": "Act like a math teacher"
        },

        {
            "role": "user",
            "content": message
        }  
    ]
)

print(response.output_text)
```

## Fetching contents of website

```python
chrome_options = Options()
chrome_options.add_argument("--headless")
chrome_options.add_argument("--disable-gpu")
driver = webdriver.Chrome(options=chrome_options)

try:
    driver.get("https://en.wikipedia.org/wiki/Neutrino")
    soup = BeautifulSoup(driver.page_source, "html.parser")

    for element in soup(["script", "style", "nav", "footer", "header"]):
        element.decompose()

    clean_text = soup.get_text(separator = "\n")
    clean_text = " ".join([line.strip() for line in clean_text.splitlines() if line.strip()])
    clean_text = clean_text[:500]

finally:
    driver.quit()

print(clean_text)
```

## Sending extracted text with prompts to OpenAI

```python
user_prompt = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.
"""

system_prompt = """
Act as a helpful assistant that analyzes the contents of a website,
and provides a short, informational summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""

client = OpenAI(
    base_url = LOCAL_BASE_URL,
    api_key = "ollama"
)

response = client.responses.create(
    model = "qwen3:0.6b",
    input = [
        {
            "role": "system",
            "content": system_prompt
        },

        {
            "role": "user",
            "content": user_prompt + " Content: " + clean_text
        }  
    ]
)

print(response.output_text)
```
