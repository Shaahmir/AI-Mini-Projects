# Eavesdropping on AI

## Importing the libraries

```python
import os
import ollama
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
```

## Loading OpenAI API Key

```python
load_dotenv(override = True)
api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    print("No API key found!")
else:
    print("API key found!")
```

## Creating OpenAI instance

```python
openai = OpenAI()
```

## Defining Prompts

```python
gpt_system_prompt = """
You are Alex. who is very argumentative and mean.
you disagree with anything in the conversation and you challenge everything, in a snarky way.
"""
```

```python
ollama_system_prompt = """
You are Blake, a very polite, courteous chatbot. You try to agree with
everything the other person says, or find common ground. If the other person is argumentative,
you try to calm them down and keep chatting.
"""
```

## Defining initial messages

```python
conversation = [
    {"speaker": "Alex", "content": "Huh, it's you again. How are you?"},
    {"speaker": "Blake", "content": "Hello! It's nice to hear from you. I'm doing very well, thank you for asking. So, what is one thing you're looking forward to right now?"}
]
```

## Calling GPT function

```python
def call_gpt():

    messages = [
        {
            "role": "system",
            "content": gpt_system_prompt
        }
    ]

    for message in conversation:

        messages.append({
            "role": "user",
            "content": f"{message['speaker']}: {message['content']}"
        })

    response = openai.responses.create(
        model = "gpt-6-luna",
        input = messages
    )

    return response.output_text
```

## Calling Ollama Function

```python
def call_ollama():

    messages = [
        {
            "role": "system",
            "content": ollama_system_prompt
        }
    ]

    for message in conversation:

        messages.append({
            "role": "user",
            "content": f"{message['speaker']}: {message['content']}"
        })

    response = ollama.chat(
        model = "llama3.2:3B",
        messages = messages
    )

    return response.message.content
```

## Eavesdropping on AI

```python
display(Markdown(f"## ChatGPT: \n{conversation[0]['content']}\n"))
display(Markdown(f"## Llama: \n{conversation[1]['content']}\n"))

for i in range(5):

    gpt_next = call_gpt()
    display(Markdown(f"## ChatGPT: \n{gpt_next}\n"))
    conversation.append({
        "speaker": "Alex",
        "content": gpt_next
    })

    ollama_next = call_ollama()
    display(Markdown(f"## Llama: \n{ollama_next}\n"))
    conversation.append({
        "speaker": "Blake",
        "content": ollama_next
    })
```
