# AI Synthetic Data Generator

## Importing the libraries

```python
import os
import json
from openai import OpenAI
from dotenv import load_dotenv
```

## Loading OpenAI API Key

```python
load_dotenv(override = True)
api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    print("No API key found!")
else:
    print("API key found!")
```

## Defining Prompts

```python
system_prompt = """
You are an expert synthetic dataset generator.

Your task is to generate realistic synthetic data.

Rules:
- Return ONLY valid JSON.
- Do NOT include markdown.
- Do NOT include backticks.
- Do NOT include explanations.
- Do NOT include comments.
- Follow the structure, field names, and data types of the example exactly.
- Generate realistic and diverse values.
- Return the requested number of records.
"""
```

```python
def user_prompt(industry, purpose, example, size):
    return f"""
    Generate a synthetic dataset for me, My requirements are given below:

    Industry: {industry}
    Purpose: {purpose}

    Here is example data showing structure:
    {example}

    Generate {size} new diverse records following the same structure.
    """
```

## Defining Tip

```python
tip = """
{
    "dataset": [
        {
            ...
        },
        {
            ...
        },
    ]
}
"""
```

## Defining Validate JSON Function

```python
def validate_json(text):
    try:
        return json.loads(text), None

    except Exception as e:
        return None, str(e)
```

## Calling OpenAI

```python
client = OpenAI()
```

```python
def generate_data(industry, purpose, example, size):

    _, err = validate_json(example)
    if err:
        raise ValueError(err + f"\nEnter JSON in this format:{tip}")

    response = client.responses.create(
        model = "gpt-5.4-mini",
        input = [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt(industry, purpose, example, size)}
        ],
        text = {"format" : { "type": "json_object" }}
    )

    return response.output_text
```

## Generating Dataset

```python
dataset = generate_data(
    "Machine Learning",
    "I want to train my K nearest neighbour model",
    """
    {
        "dataset": [
            {
                "weight": 150,
                "sweetness": 7,
                "label": "Apple"
            },
            {
                "weight": 170,
                "sweetness": 6,
                "label": "Apple"
            }
        ]
    }
    """,
    "30"
)
```
